```python
state_data = (df.groupby("Province/State")[["Confirmed","Deaths","Recovered","Active"]].sum().reset_index())
```

```python
state_data.head()
```

```python
import numpy as np
state_data["Death_Rate"] = state_data["Deaths"]/state_data["Confirmed"].replace(0,np.nan)*100
state_data["Recovery_Rate"] = (state_data["Recovered"]/state_data["Confirmed"]).replace(0,np.nan)*100

```

```python
state_data.head()
```

```python
state_data["Death_Rate"] = (
    state_data["Deaths"] / state_data["Confirmed"] * 100
)

state_data["Death_Rate"] = state_data["Death_Rate"].round(2)

print(state_data.head())
```

```python
import matplotlib.pyplot as plt

# State-wise confirmed cases
state_data = (
    df.groupby("State/UnionTerritory")["Confirmed"]
      .max()
      .reset_index()
)

# Highest cases first
state_data = state_data.sort_values(
    "Confirmed",
    ascending=False
)

# Bar chart
plt.figure(figsize=(12, 6))

plt.bar(
    state_data["State/UnionTerritory"],
    state_data["Confirmed"]
)

plt.title("Total Confirmed Cases by State")
plt.xlabel("State / Union Territory")
plt.ylabel("Confirmed Cases")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
